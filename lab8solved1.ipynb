## 1. Basic Conversation
**Exercise:** Create a simple chatbot that can answer basic questions about a given topic (e.g., history, technology).  
**Parameters to explore:** `temperature`, `max_tokens`, `top_p`, `frequency_penalty`, `presence_penalty`, `n`, `stop`.

Comment what happen when you change the parameters 
(read documentation!)

```python
import openai
import os



# Set your OpenAI API Key here (Not recommended for production)
client = openai.OpenAI(api_key=")

def ask_chatbot(prompt, temperature=0.7, max_tokens=150, top_p=1.0, frequency_penalty=0, presence_penalty=0, n=1, stop=None):
    """
    Function to interact with OpenAI's GPT-3.5-turbo chatbot using the latest OpenAI API.
    
    Parameters:
    - prompt (str): User input question.
    - temperature (float): Controls randomness (0.0 - 1.0).
    - max_tokens (int): Limits response length.
    - top_p (float): Controls diversity via nucleus sampling (0.0 - 1.0).
    - frequency_penalty (float): Reduces repetition (-2.0 to 2.0).
    - presence_penalty (float): Encourages new topics (-2.0 to 2.0).
    - n (int): Number of responses to generate.
    - stop (list or str): Stops generation when specific text is encountered.
    
    Returns:
    - str: Chatbot response.
    """
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a knowledgeable assistant that provides concise and accurate answers."},
            {"role": "user", "content": prompt}
        ],
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p,
        frequency_penalty=frequency_penalty,
        presence_penalty=presence_penalty,
        n=n,
        stop=stop
    )
    return response.choices[0].message.content

# Example usage
if __name__ == "__main__":
    question = "Tell me about the history of artificial intelligence."
    answer = ask_chatbot(question)
    print("Chatbot response:", answer)
```

## 2. Summarization
**Exercise:** Write a script that takes a long text input and summarizes it into a few sentences.  
**Parameters to explore:** `temperature`, `max_tokens`, `top_p`, `frequency_penalty`, `presence_penalty`, `best_of`, `logprobs`.

Comment what happen when you change the parameters 
(read documentation!)

```python
import openai


def summarize_text(text, temperature=0.7, max_tokens=150, top_p=1.0, frequency_penalty=0, presence_penalty=0):
    """
    Function to summarize a long text input using OpenAI's GPT-3.5-turbo model.
    
    Parameters:
    - text (str): The long input text to summarize.
    - temperature (float): Controls randomness (0.0 - 1.0).
    - max_tokens (int): Limits response length.
    - top_p (float): Controls diversity via nucleus sampling (0.0 - 1.0).
    - frequency_penalty (float): Reduces repetition (-2.0 to 2.0).
    - presence_penalty (float): Encourages new topics (-2.0 to 2.0).
    
    Returns:
    - str: Summarized text.
    """
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are an AI that provides concise and accurate text summaries."},
            {"role": "user", "content": f"Summarize the following text: {text}"}
        ],
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p,
        frequency_penalty=frequency_penalty,
        presence_penalty=presence_penalty
    )
    return response.choices[0].message.content

# Example usage
if __name__ == "__main__":
    long_text = """Artificial intelligence (AI) has rapidly evolved over the years, with applications ranging from natural language processing to robotics. Early AI research focused on symbolic reasoning and expert systems, while modern AI emphasizes deep learning and neural networks. As AI continues to advance, ethical considerations and regulatory frameworks become increasingly important in ensuring responsible use."""
    summary = summarize_text(long_text)
    print("Summarized text:", summary)
```

## 3. Translation
**Exercise:** Develop a tool that translates text from one language to another using the API.  
**Parameters to explore:** `temperature`, `max_tokens`, `top_p`, `frequency_penalty`, `presence_penalty`, `echo`, `logit_bias`.

(read documentation!)

```python
import openai

def translate_text(text, target_language, temperature=0.7, max_tokens=150, top_p=1.0, frequency_penalty=0, presence_penalty=0):
    """
    Function to translate text from one language to another using OpenAI's GPT-3.5-turbo model.
    
    Parameters:
    - text (str): The input text to translate.
    - target_language (str): The language to translate the text into.
    - temperature (float): Controls randomness (0.0 - 1.0).
    - max_tokens (int): Limits response length.
    - top_p (float): Controls diversity via nucleus sampling (0.0 - 1.0).
    - frequency_penalty (float): Reduces repetition (-2.0 to 2.0).
    - presence_penalty (float): Encourages new topics (-2.0 to 2.0).
    
    Returns:
    - str: Translated text.
    """
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a translation assistant that provides accurate translations."},
            {"role": "user", "content": f"Translate the following text into {target_language}: {text}"}
        ],
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p,
        frequency_penalty=frequency_penalty,
        presence_penalty=presence_penalty
    )
    return response.choices[0].message.content

# Example usage
if __name__ == "__main__":
    input_text = "Artificial intelligence is transforming the world."
    target_lang = "Spanish"
    translated_text = translate_text(input_text, target_lang)
    print("Translated text:", translated_text)

```

## 4. Sentiment Analysis
**Exercise:** Implement a sentiment analysis tool that determines the sentiment of a given text (positive, negative, neutral).  
**Parameters to explore:** `temperature`, `max_tokens`, `top_p`, `frequency_penalty`, `presence_penalty`, `n`, `logprobs`.

Comment what happen when you change the parameters 
(read documentation!)

```python
import openai


def analyze_sentiment(text, temperature=0.7, max_tokens=50, top_p=1.0, frequency_penalty=0, presence_penalty=0, n=1):
    """
    Function to analyze the sentiment of a given text using OpenAI's GPT-3.5-turbo model.
    
    Parameters:
    - text (str): The input text to analyze.
    - temperature (float): Controls randomness (0.0 - 1.0).
    - max_tokens (int): Limits response length.
    - top_p (float): Controls diversity via nucleus sampling (0.0 - 1.0).
    - frequency_penalty (float): Reduces repetition (-2.0 to 2.0).
    - presence_penalty (float): Encourages new topics (-2.0 to 2.0).
    - n (int): Number of responses to generate.
    
    Returns:
    - str: The sentiment classification (Positive, Negative, Neutral).
    """
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a sentiment analysis assistant that classifies text as Positive, Negative, or Neutral."},
            {"role": "user", "content": f"Determine the sentiment of the following text: {text}"}
        ],
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p,
        frequency_penalty=frequency_penalty,
        presence_penalty=presence_penalty,
        n=n
    )
    return response.choices[0].message.content.strip()

# Example usage
if __name__ == "__main__":
    input_text = "I absolutely love this product! It's amazing and exceeded my expectations."
    sentiment = analyze_sentiment(input_text)
    print("Sentiment:", sentiment)
```

## 5. Text Completion
**Exercise:** Create a text completion application that generates text based on an initial prompt.  
**Parameters to explore:** `temperature`, `max_tokens`, `top_p`, `frequency_penalty`, `presence_penalty`, `stop`, `best_of`.

Comment what happen when you change the parameters 
(read documentation!)

```python
import openai

def complete_text(prompt, temperature=0.7, max_tokens=100, top_p=1.0, frequency_penalty=0, presence_penalty=0, stop=None, best_of=1):
    """
    Function to generate text completion based on an initial prompt using OpenAI's GPT-3.5-turbo model.
    
    Parameters:
    - prompt (str): The initial text input for generation.
    - temperature (float): Controls randomness (0.0 - 1.0).
    - max_tokens (int): Limits response length.
    - top_p (float): Controls diversity via nucleus sampling (0.0 - 1.0).
    - frequency_penalty (float): Reduces repetition (-2.0 to 2.0).
    - presence_penalty (float): Encourages new topics (-2.0 to 2.0).
    - stop (list or str): Stops generation when specific text is encountered.
    - best_of (int): Generates multiple completions and returns the best one.
    
    Returns:
    - str: Completed text.
    """
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a creative text completion assistant."},
            {"role": "user", "content": prompt}
        ],
        temperature=temperature,
        max_tokens=max_tokens,
        top_p=top_p,
        frequency_penalty=frequency_penalty,
        presence_penalty=presence_penalty,
        stop=stop
    )
    return response.choices[0].message.content.strip()

# Example usage
if __name__ == "__main__":
    initial_prompt = "Once upon a time in a distant land, a young adventurer set out on a journey to"
    completion = complete_text(initial_prompt)
    print("Completed text:", completion)
```
